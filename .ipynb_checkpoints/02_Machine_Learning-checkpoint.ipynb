```python
# ==========================================
# ML CHUNK 1 : LOAD CLEAN DATASET
# ==========================================

import pandas as pd

# Load the cleaned dataset
df = pd.read_csv("Clean_Flight_Data.csv")

# Display the first five records
df.head()
```

```python
# Dataset dimensions
print("Dataset Shape:", df.shape)

# Column names
print(df.columns)

# Data types
df.info()
```

```python
# ==========================================
# Convert Duration to Minutes
# ==========================================

def convert_duration(duration):
    hours = 0
    minutes = 0

    if 'h' in duration:
        hours = int(duration.split('h')[0])

    if 'm' in duration:
        minutes = int(duration.split('m')[0].split()[-1])

    return hours * 60 + minutes

df['Duration_Minutes'] = df['Duration'].apply(convert_duration)
```

```python
# ==========================================
# Extract Arrival Hour
# ==========================================

df['Arrival_Hour'] = pd.to_datetime(
    df['Arrival_Time'].str[:5],
    format='%H:%M'
).dt.hour
```

```python
# ==========================================
# Drop Unnecessary Columns
# ==========================================

df.drop(
    columns=[
        'Date_of_Journey',
        'Dep_Time',
        'Arrival_Time',
        'Duration',
        'Route'
    ],
    inplace=True
)
```

```python
df.info()
```

```python
# ==========================================
# ML CHUNK 2 : FEATURE SELECTION
# ==========================================

# Define the target variable
y = df["Price"]

# Define the feature variables
X = df.drop(columns=["Price"])

# Display feature matrix
X.head()
```

```python
print("Number of Features:", X.shape[1])
print("Target Variable:", y.name)
```

```python
df.info()
```

```python
# ==========================================
# ML CHUNK 3 : ENCODE CATEGORICAL VARIABLES
# ==========================================

# Apply One-Hot Encoding
df_encoded = pd.get_dummies(
    df,
    columns=[
        'Airline',
        'Source',
        'Destination',
        'Total_Stops',
        'Additional_Info',
        'Journey_Month',
        'Journey_Day'
    ],
    drop_first=True
)

# Display the first five rows
df_encoded.head()
```

```python
# Check the new dataset structure
df_encoded.info()
```

```python
# ==========================================
# DEFINE FEATURES AND TARGET
# ==========================================

# Target variable
y = df_encoded['Price']

# Feature variables
X = df_encoded.drop(columns=['Price'])

print("Feature Matrix Shape:", X.shape)
print("Target Shape:", y.shape)
```

```python
# ==========================================
# ML CHUNK 4 : TRAIN-TEST SPLIT
# ==========================================

from sklearn.model_selection import train_test_split

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Display the shapes
print("Training Features:", X_train.shape)
print("Testing Features :", X_test.shape)

print("Training Target  :", y_train.shape)
print("Testing Target   :", y_test.shape)
```

## ML Chunk 4: Train-Test Split

### Objective

Split the dataset into training and testing sets to evaluate the model on unseen data.

### Why?

Machine learning models should be assessed on data they have not seen during training. An 80:20 split provides sufficient data for training while reserving a representative portion for evaluation.

### Outcome

- **Training Set:** Used to train the regression models.
- **Testing Set:** Used to evaluate prediction accuracy and compare model performance.

```python
# ==========================================
# ML CHUNK 5 : IMPORT LIBRARIES
# ==========================================

from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import numpy as np
```

```python
# ==========================================
# TRAIN LINEAR REGRESSION MODEL
# ==========================================

# Create the model
lr_model = LinearRegression()

# Train the model
lr_model.fit(X_train, y_train)
```

```python
# ==========================================
# MAKE PREDICTIONS
# ==========================================

# Predict ticket prices
y_pred_lr = lr_model.predict(X_test)
```

```python
# ==========================================
# MODEL EVALUATION
# ==========================================

# Calculate evaluation metrics
mae = mean_absolute_error(y_test, y_pred_lr)

mse = mean_squared_error(y_test, y_pred_lr)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred_lr)

# Display results
print("Linear Regression Performance")
print("-" * 35)

print(f"MAE  : {mae:.2f}")
print(f"MSE  : {mse:.2f}")
print(f"RMSE : {rmse:.2f}")
print(f"R² Score : {r2:.4f}")
```

## Linear Regression Results

### Observation

- The Linear Regression model achieved an **R² Score of 0.6817**, indicating that it explains approximately **68.17%** of the variation in airline ticket prices.
- The model recorded a **Mean Absolute Error (MAE) of ₹1,740.33**, meaning that the predicted ticket prices differ from the actual prices by about ₹1,740 on average.
- The **Root Mean Squared Error (RMSE) of ₹2,577.11** suggests that while the model performs reasonably well, some flights have larger prediction errors.
- Overall, the model demonstrates a moderate ability to predict airline ticket prices using the selected flight characteristics.

### Business Insight

The results indicate that variables such as **airline, source, destination, journey month, duration, number of stops, and departure hour** collectively explain a significant portion of ticket price variation. However, nearly **32% of the variation remains unexplained**, suggesting that airline pricing is also influenced by factors not included in the dataset, such as booking lead time, seat availability, seasonal demand, and competitor pricing.

### Conclusion

Linear Regression provides a strong baseline model for airline ticket price prediction. However, because airline pricing is influenced by complex and non-linear relationships, more advanced algorithms such as **Decision Tree** and **Random Forest** are expected to improve predictive accuracy.

```python
# ==========================================
# ACTUAL VS PREDICTED
# ==========================================

comparison = pd.DataFrame({
    "Actual Price": y_test.values,
    "Predicted Price": np.round(y_pred_lr, 2)
})

comparison.head(10)
```

```python
# ==========================================
# ACTUAL VS PREDICTED SCATTER PLOT
# ==========================================
import matplotlib.pyplot as plt
plt.figure(figsize=(8,6))

plt.scatter(y_test, y_pred_lr, alpha=0.6)

plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Actual vs Predicted Ticket Prices (Linear Regression)")

plt.show()
```

```python
# ==========================================
# ML CHUNK 6 : IMPORT LIBRARIES
# ==========================================

from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
import numpy as np
```

```python
# ==========================================
# TRAIN DECISION TREE REGRESSOR
# ==========================================

dt_model = DecisionTreeRegressor(random_state=42)

dt_model.fit(X_train, y_train)
```

```python
# ==========================================
# MAKE PREDICTIONS
# ==========================================

y_pred_dt = dt_model.predict(X_test)
```

```python
# ==========================================
# MODEL EVALUATION
# ==========================================

mae_dt = mean_absolute_error(y_test, y_pred_dt)

mse_dt = mean_squared_error(y_test, y_pred_dt)

rmse_dt = np.sqrt(mse_dt)

r2_dt = r2_score(y_test, y_pred_dt)

print("Decision Tree Performance")
print("-" * 35)

print(f"MAE      : {mae_dt:.2f}")
print(f"MSE      : {mse_dt:.2f}")
print(f"RMSE     : {rmse_dt:.2f}")
print(f"R² Score : {r2_dt:.4f}")
```

## Decision Tree Regression Results

### Observation

- The Decision Tree Regressor achieved an **R² Score of 0.7783**, explaining approximately **77.83%** of the variation in airline ticket prices.
- The model recorded a **Mean Absolute Error (MAE) of ₹988.78**, indicating that predictions differ from the actual ticket prices by about ₹989 on average.
- The **Root Mean Squared Error (RMSE) of ₹2,150.57** is lower than that of the Linear Regression model, demonstrating improved prediction accuracy.
- Compared with the baseline model, the Decision Tree achieved lower prediction errors and a higher explanatory power.

### Business Insight

The improved performance indicates that airline ticket prices are influenced by complex, non-linear relationships among variables such as airline, route, journey duration, number of stops, and travel period. The Decision Tree is better able to capture these interactions, resulting in more accurate price predictions.

### Conclusion

The Decision Tree Regressor outperformed the Linear Regression model across all evaluation metrics, making it a stronger candidate for predicting airline ticket prices. However, because Decision Trees can overfit the training data, it is important to evaluate a more robust ensemble model, such as Random Forest, before selecting the final predictive model.

```python
# ==========================================
# ACTUAL VS PREDICTED
# ==========================================

comparison_dt = pd.DataFrame({
    "Actual Price": y_test.values,
    "Predicted Price": np.round(y_pred_dt, 2)
})

comparison_dt.head(10)
```

```python
# ==========================================
# ACTUAL VS PREDICTED SCATTER PLOT
# ==========================================

plt.figure(figsize=(8,6))

plt.scatter(y_test, y_pred_dt, alpha=0.6)

plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Actual vs Predicted Ticket Prices (Decision Tree)")

plt.show()
```

```python
# ==========================================
# ML CHUNK 7 : IMPORT LIBRARIES
# ==========================================

from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import numpy as np
```

```python
# ==========================================
# TRAIN RANDOM FOREST REGRESSOR
# ==========================================

rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)
```

```python
# ==========================================
# MAKE PREDICTIONS
# ==========================================

y_pred_rf = rf_model.predict(X_test)
```

```python
# ==========================================
# MODEL EVALUATION
# ==========================================

mae_rf = mean_absolute_error(y_test, y_pred_rf)

mse_rf = mean_squared_error(y_test, y_pred_rf)

rmse_rf = np.sqrt(mse_rf)

r2_rf = r2_score(y_test, y_pred_rf)

print("Random Forest Performance")
print("-" * 35)

print(f"MAE      : {mae_rf:.2f}")
print(f"MSE      : {mse_rf:.2f}")
print(f"RMSE     : {rmse_rf:.2f}")
print(f"R² Score : {r2_rf:.4f}")
```

## Random Forest Regression Results

### Observation

- The Random Forest Regressor achieved an **R² Score of 0.8284**, explaining approximately **82.84%** of the variation in airline ticket prices.
- The model recorded a **Mean Absolute Error (MAE) of ₹903.83**, meaning that predicted ticket prices differ from the actual prices by approximately ₹904 on average.
- The **Root Mean Squared Error (RMSE) of ₹1,892.43** is the lowest among all evaluated models, indicating better prediction accuracy and fewer large prediction errors.
- Compared with the Linear Regression and Decision Tree models, the Random Forest model consistently achieved superior performance across all evaluation metrics.

### Business Insight

The Random Forest model effectively captures the complex relationships between airline, source, destination, journey month, duration, number of stops, and other flight characteristics. This makes it well suited for predicting airline ticket prices and demonstrates that ticket pricing is influenced by multiple interacting factors rather than simple linear relationships.

### Conclusion

Among all evaluated models, the **Random Forest Regressor** delivered the highest predictive performance and is selected as the final model for airline ticket price prediction. Its higher accuracy makes it a reliable tool for supporting data-driven pricing decisions and estimating ticket prices based on flight characteristics.

```python
# ==========================================
# ACTUAL VS PREDICTED
# ==========================================

comparison_rf = pd.DataFrame({
    "Actual Price": y_test.values,
    "Predicted Price": np.round(y_pred_rf, 2)
})

comparison_rf.head(10)
```

```python
# ==========================================
# ML CHUNK 8 : MODEL COMPARISON
# ==========================================

model_comparison = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Decision Tree",
        "Random Forest"
    ],
    "MAE": [
        mae,
        mae_dt,
        mae_rf
    ],
    "RMSE": [
        rmse,
        rmse_dt,
        rmse_rf
    ],
    "R² Score": [
        r2,
        r2_dt,
        r2_rf
    ]
})

# Round values for better readability
model_comparison = model_comparison.round(2)

model_comparison
```

## Model Comparison

### Observation

- Three regression models were evaluated using the same training and testing datasets.
- Random Forest achieved the highest **R² Score (0.8284)** and the lowest **MAE (₹903.83)** and **RMSE (₹1,892.43)**.
- Decision Tree performed better than Linear Regression but did not outperform Random Forest.
- Linear Regression served as a useful baseline model but was less effective at capturing the complex relationships in the dataset.

### Business Insight

The comparison demonstrates that ensemble learning techniques, such as Random Forest, provide more accurate predictions for airline ticket prices than simpler regression models. This suggests that airline pricing is driven by non-linear interactions among multiple flight characteristics.

### Conclusion

Based on the evaluation metrics, **Random Forest** is selected as the final model for airline ticket price prediction because it provides the best balance of prediction accuracy and model reliability.
